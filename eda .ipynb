```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("amazon sale report.csv", low_memory=False)
df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
```

```python
df = df.drop(columns=['Unnamed: 22'], errors='ignore')
df = df.drop(columns=['promotion-ids'], errors='ignore')
```

```python
df['Amount'] = df['Amount'].fillna(0)
df['currency'] = df['currency'].fillna('Unknown')
df['Courier Status'] = df['Courier Status'].fillna('Unknown')
df['fulfilled-by'] = df['fulfilled-by'].fillna('Unknown')
```

```python
df = df.dropna(subset=['ship-city', 'ship-state'])
```

```python
df.isnull().sum()
df.info()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
df['Status'].value_counts().plot(kind='bar')
plt.title("Order Status Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df['Category'].value_counts().plot(kind='bar')
plt.title("Product Categories Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df['ship-state'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 States by Orders")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df.groupby('Category')['Amount'].sum().sort_values(ascending=False).plot(kind='bar')
plt.title("Revenue by Category")
plt.show()
```
